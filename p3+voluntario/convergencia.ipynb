```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Utilizar el estilo ggplot de Matplotlib
plt.style.use('ggplot')

# Función para leer el archivo y obtener los datos
def read_data(file_name):
    N_exp = []
    Transmission = []
    with open(file_name, 'r') as file:
        next(file)  # Saltar la primera línea si tiene encabezados
        for line in file:
            data = line.strip().split(',')
            N_exp.append(int(data[1].strip()))
            Transmission.append(float(data[2].strip()))
    return N_exp, Transmission

# Función para graficar la convergencia del coeficiente de transmisión
def plot_convergence(file_name):
    N_exp, Transmission = read_data(file_name)
    
    plt.figure(figsize=(10, 6))
    plt.plot(N_exp, Transmission, marker='.', linestyle='-', color='b', label='λ=0.7')
    plt.title('Convergencia del Coeficiente de Transmisión', fontsize=16)  # Tamaño de fuente del título
    plt.xlabel('Número de experimentos', fontsize=14)  # Tamaño de fuente del eje x
    plt.ylabel('Coeficiente de transmisión', fontsize=14)  # Tamaño de fuente del eje y
    plt.grid(True)
    plt.legend(fontsize=20)  # Tamaño de fuente de la leyenda
    plt.tight_layout()
    plt.show()

# Nombre del archivo de entrada
input_file = './DESKTOP/lambda=0.7.txt'  # Reemplaza con la ruta a tu archivo

# Generar la gráfica de convergencia
plot_convergence(input_file)
```
