```python
import matplotlib.pyplot as plt
import numpy as np
import csv
```

```python
x_values = []
y_values = []
with open("magnetizacion.txt", 'r') as file:
        csv_reader = csv.reader(file)
        next(csv_reader)  # Skip the header row if it exists
        for row in csv_reader:
            x_values.append(float(row[0]))  # Convert X values to float
            y_values.append(float(row[1]))  # Convert Y values to float

# Plot the data
plt.plot(x_values, y_values)
plt.xlabel('Temperatura')
plt.ylabel('Magnetizacion')
plt.title('Modelo de Ising - Magnetizacion vs Temperatura')
plt.grid(True)
plt.show()
```
